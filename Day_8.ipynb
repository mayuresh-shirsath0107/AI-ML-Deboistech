Q1 — Even or Odd Counter

20 min · loops, conditionals, comprehensions  
You're given a hardcoded list of numbers:   
numbers = [12, 7, 9, 4, 22, 15, 8, 3, 30, 11]

Write a script that:

Loops through the list and prints whether each number is even or odd                
Counts how many are even and how many are odd, using simple counters             
Uses a comprehension to build a new list containing only the even numbers       
Prints the final counts and the even-numbers list

```python
numbers = [12, 7, 9, 4, 22, 15, 8, 3, 30, 11]

even =[] #stores the even numbers from the list
even_count = 0 #count of even numbers
odd_count = 0 #count of odd numbers

for num in numbers: #execute on every number from the list
  if num % 2 ==0: #sorting even numbers
    print(f'{num} is even')
    even.append(num) #stored the even number in even[]
    even_count += 1 #increment the count of even numbers
  else:
    print(f'{num} is odd')
    odd_count += 1 #increment the count of odd numbers
print(f'counts of even numbers are {even_count}')
print(f'counts of odd numbers are {odd_count}')
print(f'even numbers are {even}')

```

Q3 — Number Guessing Stats                
20 min · random module, loops, math module    
Simulate the computer guessing a random number between 1 and 50, ten times, using the random module.                             

Use random.randint(1, 50) to generate each of the 10 "guesses"                          
Store all 10 guesses in a list, printing each one as it's generated                       
After the loop, use the math module to print the highest guess (math.floor won't be needed here — just use max()) and the average guess
rounded to 2 decimal places



```python
import random
import math


guesses=[] #stores guesses

for i in range(10): # loop will execute 10 times
  guess = random.randint(1,50) #generate random numbers from 1 to 50
  guesses.append(guess) #generated numbers will store in guesses variable
  print(f'Guess {i+1}: {guess}') #print which guess it was and what is the guess

print(f'highest guess is {max(guesses)}') #prints the Maximum value from the guesses 
print(f'average guess is {round(sum(guesses)/len(guesses),2)}') #prints average of guesses

```

Q4 — Student Class                           
                                            
30 min · classes, __init__, type hints, one dunder method                              
Design a Student class:                      
                                            
- __init__(self, name: str, marks: list) — both type-hinted instance attributes             
- A method average(self) -> float that returns the average of the student's marks            
- __str__ — returns a readable one-line summary, e.g. "Sakshi: average 82.5"
- Then create a hardcoded list of 3 Student objects, print each one (this should use your __str__ automatically), and print the name of
whichever student has the highest average.

```python
class Student: #created class Student
  def __init__(self, name:str , marks:list): #initializing arguments 
    self.name:str = name
    self.marks:list = marks

  def average(self)-> float: #creating average function
    return sum(self.marks)/len(self.marks) #calculating average of marks

  def __str__(self): #creating module which return the name and average of marks of students
    return f'{self.name}: average {self.average()}'
    


students = [   #creating objects
    Student("Varun", [80,79,67,91,55]),
    Student("Mayuresh",[91,75,88,69,78]),
    Student("Raj", [66,76,81,71,55])]

for student in students: #printing every object 
  print(student)

highest_avg = max(students, key=lambda Student: student.average()) #returns highest average from the all students 
print(f"The Highest Average is :  {highest_avg}")
```
