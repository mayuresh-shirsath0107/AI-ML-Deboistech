```python
import numpy as np
numbers =[10,20,30] #Python list
'''list are unorganized datatype in which we can use multiple data types '''

number=np.array([10,20,30]) #Numpy array
'''arrays are organized datatset in which only same type of datatypes can be use'''
```

```python
print(numbers)
print(number)
```

```python
numbers = [10 ,20 ,30] #list
result=[]

for n in numbers:
  result.append(n*2)
result
'''Execution flow----
in for loop , every number in list "numbers" will multiply with 2 amd the result will be append in empty result =[] variable 
at last.... we call the updated result  '''
```

```python
numbers = np.array([10,20,30]) #arrays
result = numbers*2
result
''' In arrays ... every element will multiply with 2 and the output will store in result...
result is actually a new array in which output is stored '''
```
