WHILE loops == this Loop run continuously as long as given condition is TRUE  
                       or  until condition is FALSE                                             

```python
spam=0
if spam<5:
  print("Hello world")
  spam=spam+1
```

```python
spam=0
while spam<5:
  print("Hello world")
  spam=spam+1
```

```python
name=""
while name!="Mayuresh":
  print("Please Type your Name:")
  name=input('>')
print("Thank you")
```

**Break== Stops the Loop when Condition is True**

```python
while True:
  print("Type your Name:")
  name=input('>')
  if name == "Mayuresh":
   break
print("Thank you")
```

**Continue == Skips the iteration when condition is True**

```python
for num in range(10):
 if num%2==0:
  continue
print(num)
```

**For Loops** == loops repeat the code block specific number of times by iterations

```python
for i in range (5):
  print("Hello")
```

**range() ==takes arguments(start,stop,step)**

```python
#range(stop)
for i in range (3):
  print("Hello")
```

```python
#range(start,stop)
for i in range (1,11):
  print(i)
```

```python
#range(start,stop,step)
for i in range (1,11,2):
  print(i)
```

GUESS THE NUMBER

```python
# This is a guess the number game.
import random
secret_number = random.randint(1, 20)
print('I am thinking of a number between 1 and 20.')
# Ask the player to guess 6 times.
for guesses_taken in range(1, 7):
    print('Take a guess.')
    guess = int(input('>'))
    if guess < secret_number:
        print('Your guess is too low.')
    elif guess > secret_number:
        print('Your guess is too high.')
    else:
        break # This condition is the correct guess!
if guess == secret_number:
    print('Good job! You got it in ' + str(guesses_taken) + ' guesses!')
else:
    print('Nope. The number was ' + str(secret_number))
```
