 StopIteration occurs when list or values are ends to iterates in iterator

```python
class Countdown:
  def __init__(self,start): #Passing dynamic arguments
    self.current = start
  def __iter__(self):# creating iterator
    return self
  def __next__(self):
    if self.current <0:
      raise StopIteration
    number = self.current #variable in which stores cuurent argument
    self.current -= 1 #when value return , it basically removes from memory
    return number
for number in Countdown(3):
  print(number)
```

```python
def count_up_to(limit):
  number = 1
  while number <= limit:
    yield number
    number += 1
for number in count_up_to(5):
  print(number)
```

# **Decoration**

```python
def say_hello():
  name =input("enter usr name :")
  print(f'Heyy {name}')

def decorator(func):

  def wrapper():
    print("who is the user that side? ")
    func()

  return wrapper

say_hello = decorator(say_hello)
say_hello()
```

Handling any arguments : *args, **kwargs

```python
def decorator(function):
  def wrapper(*args,**kwargs):
    result = function(*args,**kwargs)
    return result
  return wrapper
@decorator
def add(a,b):
  return a+b
add(10,20)
```

Proper decorator example

```python
def decorator(function): # (function) is actually a base function we want to decorate .....eg.add()
  def wrapper(*args,**kwargs):  #args and kwargs sllos the function to accept any arguments (numbers and key words )
    print("And the addition is ....") #Before the function

    result = function(*args,**kwargs)
    print("Is the answer correct.?") #After the function #uses original function again in result
    return result
    print("Is the answer correct.?") #After the function
  return wrapper


@decorator #Its just a shortcut for decoration call
def add(a,b):
  return a+b
add(10,20)
```
