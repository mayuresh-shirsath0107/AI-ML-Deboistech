# **Integers**

```python
#Arithmetic Operations
a = 55
b = 74
print(a + b)
print(a - b)
```

```python

# Exponent
a = 8
b = 4
print(a ** b)
```

```python
# Modulus
a = 99
b = 8
print(a % b)
```

```python
# Average
a = 66
b = 95
c = 89
d = 56
e = 76
avg = (a + b + c + d + e) / 5
print(avg)
```

# **String**

```python
# Index
name = "Python"
print(name[0])
print(name[4])
```

```python
# Slice
print(name[0:2])
print(name[:5])
```

```python
# Methods
name = "Mayuresh"
print(name.upper())
print(name.lower())
```

```python
#Concatination and Replication
a = "Hello "
b = "World"
print(a + b)
print(a * 3)
```

# **List**

```python
# Index
nums = [10, 15, 20, 25, 30]
print(nums[3])
print(nums[-1])
```

```python
# Sort
lst = [43, 21, 12, 87, 32]
lst.sort()
print(lst)
```

```python
#Append
num= [10, 15, 20]
num.append(25)
print(num)
```

```python
# Remove
num = [15, 25, 35, 45]
num.remove(25)# revove speific value
print(num)
x = num.pop()#remove last value in the list
print(x)
print(num)
```

# **Dictionaries**

```python
# Access
student = {
    "name": "Mayuresh",
    "age": 22,
    "Roll": 32
}
print(student["name"])
print(student["Roll"])
```

```python
# Update
student = {
    "name": "Mayuresh",
    "age": 22,
    "Roll": 32
}
student["age"] = 23
print(student)
```

```python
# Add
student = {
    "name": "Mayuresh",
    "age": 22,
    "Roll": 32
}
student["batch"] = "AI/ML"
print(student)
```
