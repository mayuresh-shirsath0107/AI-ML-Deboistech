**TypeError**

```python
try:
    num = int(input("Enter a number: "))
    text = input("Enter some text: ")

    result = num + text
    print("Result =", result)

except TypeError:
    print("TypeError: Cannot add a number and a string.")
```

**ZeroDivisionError**

```python
try:
    num1 = int(input("Enter numerator: "))
    num2 = int(input("Enter denominator: "))

    result = num1 / num2
    print("Result =", result)

except ZeroDivisionError:
    print("ZeroDivisionError: Division by zero is not allowed.")
```

**ValueError**

```python
try:
    num = int(input("Enter a nnumber: "))
    print("You entered:", num)

except ValueError:
    print("ValueError: Please enter a valid number.")
```

**IndexError**

```python
try:
    num = [10, 20, 30, 40, 50]

    i = int(input("Enter index (0-4): "))
    print("Element =", num[i])

except IndexError:
    print("IndexError: Index is out of range.")
```
