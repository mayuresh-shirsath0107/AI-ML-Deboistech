Exception handling : used to continue the program execution even if code throws an error

```python
def spam(div):
  try: #if error occured , program will execute except condition
    return 42 / div
  except ZeroDivisionError: #skips when if proram executes successfully
    print("Error: Invalid argument")

print(spam(2))
print(spam(12))
print(spam(0))
print(spam(3))
```

# **Mini Project**

```python
#this is animation program
import time, sys
indent = 0 # How many spaces to indent
indent_increasing = True # Whether the indentation is increasing or not
try:
    while True: # The main program loop
        print(' ' * indent, end='')
        print('********')
        time.sleep(0.1) # Pause for 1/10th of a second.
        if indent_increasing:
            # Increase the number of spaces:
            indent = indent + 1
    if indent == 20:
        # Change direction:
        indent_increasing = False
    else:
        # Decrease the number of spaces:
        indent = indent - 1
    if indent == 0:
        # Change direction:
        indent_increasing = True
except KeyboardInterrupt:
    sys.exit()
```
